```python
import matplotlib.pyplot as plt
```

Array Untuk Menyimpan Nilai dan Input 10 Nilai

```python
nilai = []

for i in range(10):
    n = int(input(f"Masukkan nilai mahasiswa ke-{i+1}: "))
    nilai.append(n)
```

Menampilkan Semua nilai

```python
print("Daftar nilai:", nilai)
```

Menghitung nilai Tertinggi, Terendah, Rata-rata

```python
tertinggi = max(nilai)
terendah = min(nilai)

print("Nilai tertinggi:", tertinggi)
print("Nilai terendah:", terendah)

rata = sum(nilai) / len(nilai)
print("Rata-rata nilai:", rata)
```

Menghitung jumlah mahasiswa lulus >60

```python
lulus = 0
tidak_lulus = 0

for n in nilai:
    if n >= 60:
        lulus += 1
    else:
        tidak_lulus += 1

print("Jumlah mahasiswa lulus:", lulus)
print("Jumlah mahasiswa tidak lulus:", tidak_lulus)
```

Grafik tinggi rendahnya nilai

```python
plt.bar(["Tertinggi", "Terendah"], [tertinggi, terendah])
plt.title("Grafik Nilai Tertinggi dan Terendah")
plt.ylabel("Nilai")
plt.show()
```

Grafik mahasiswa lulus, tidak lulus

```python
plt.bar(["Lulus", "Tidak Lulus"], [lulus, tidak_lulus])
plt.title("Grafik Data Kelulusan")
plt.ylabel("Jumlah Mahasiswa")
plt.show()
```
